```python
def histogramas(columns, df,nombre,x=False):
    for col in columns:
        if pd.api.types.is_numeric_dtype(df[col]):
            sd = np.std(fdrec[col])
            mn = np.mean(fdrec[col])
            plt.figure(figsize=(10, 6))
            
            plt.hist(df[col].dropna(), bins=20, edgecolor='black', color='#81d4fa')
            if x == True:
                plt.axvline(mn + sd, color='#f08080', linestyle='--', label=f'1 Desv. Est.{round((mn + sd),2)}')
                plt.axvline(mn - sd, color='#f08080', linestyle='--', label=f'- 1 Desv. Est.{round((mn - sd),2)}')
                plt.axvline(mn - 2*sd, color='#a5d6a7', linestyle='--', label=f'- 2 Desv. Est.{round((mn - 2*sd),2)}')
                plt.axvline(mn + 2*sd, color='#a5d6a7', linestyle='--', label=f'+ 2 Desv. Est.{round((mn + 2*sd),2)}')
                plt.axvline(mn, color='green', linestyle='-', label=f'Media {round(mn,2)}')


            plt.xlabel(col)
            plt.ylabel('Frecuency')
            plt.legend()
            plt.savefig(f"../images/{nombre}.png")
            plt.show()
            
        else:
            print(f"La columna '{col}' no es numérica y se omite.")
```

```python
#"xmax","dxmax"
histogramas(columns=["xmax"],df=fdrec,nombre="xmaxhist",x=False)
```

```python
#distribuciones
def mapa_correlacion(df):
    # Calcular la matriz de correlación
    correlacion = df.corr(numeric_only=True)

    # Crear el mapa de calor
    plt.figure(figsize=(10, 8))
    sns.heatmap(correlacion, annot=True, cmap='coolwarm', center=0, fmt=".2f")
    plt.title('Matriz de Correlación')
    plt.tight_layout()
    plt.savefig("../images/heathist.png")
    plt.show()
```

```python
# Agregar 
import pandas as pd
import numpy as np
import random
# Supongamos que tienes un DataFrame con una columna 'Xmax'
# df = pd.read_csv("tus_datos.csv")

# Ejemplo de generación de muestra aleatoria si aún no tienes el archivo:
# df = pd.DataFrame({'Xmax': np.random.uniform(680, 1050, 1000)})

def assign_particle(xmax):
    # Definimos las probabilidades basadas en rangos
    random.seed(2)
    if xmax < 720:
        prob_proton = 0.01
    elif xmax < 740:
        prob_proton = 0.05
    elif xmax < 760:
        prob_proton = 0.15
    elif xmax < 780:
        prob_proton = 0.30
    elif xmax < 800:
        prob_proton = 0.50
    elif xmax < 820:
        prob_proton = 0.70
    elif xmax < 840:
        prob_proton = 0.85
    elif xmax < 880:
        prob_proton = 0.95
    else:
        prob_proton = 0.99
    
    # Asignamos clase basada en una tirada aleatoria
    return 'proton' if np.random.rand() < prob_proton else 'iron'

# Asignar nueva columna
fdrec_particle  = fdrec.copy()
fdrec_particle["particle_type"] = fdrec_particle["xmax"].apply(assign_particle)

# Opcional: guardar nuevo archivo
fdrec_particle.to_csv("fdrec_clasificados.csv",index=False)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
import pandas as pd

def modelo_arbol_decision(df, target_col='particle_type', max_depth=3, test_size=0.3, random_state=42):


    # Verificar si el target existe
    if target_col not in df.columns:
        raise ValueError(f"La columna objetivo '{target_col}' no está en el DataFrame.")

    # Separar características (X) y etiquetas (y)
    y = df[target_col].astype(str)
    X = df.drop(columns=[target_col])

    # Dividir los datos
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, random_state=random_state
    )

    # Modelo
    model = DecisionTreeClassifier(max_depth=max_depth, random_state=random_state)
    model.fit(X_train, y_train)

    # Predicciones
    y_pred = model.predict(X_test)

    # Reporte
    report = classification_report(y_test, y_pred)

    # DataFrame de resultados
    resultado_df = X_test.copy()
    resultado_df['true_label'] = y_test.values
    resultado_df['predicted'] = y_pred

    return model, report, resultado_df
```

```python
model=fdrec_particle[["theta","phi","dec","totalEnergy","dtotalEnergy","calEnergy","dcalEnergy","xmax"
                        ,"heightXmax","dEdXmax","particle_type"]]
modelo, reporte, resultado_df=modelo_arbol_decision(df=model, max_depth=3, test_size=0.20, random_state=42)
```
